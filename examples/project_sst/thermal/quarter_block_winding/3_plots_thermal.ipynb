# Configurable COMSOL 3D plot exports

Choose the view, plot groups, units, and color ranges, then call the reusable generic export helper.


```python
from pathlib import Path
import mphsweepkit as msk

# Choose one orthographic view: 'xy', 'yz', or 'xz'.
view_plane = 'xy'

# Add, remove, or change plot groups here. A feature is only needed when
# changing that plot's unit or color range.
plots = [
    {
        'plot_group': 'Thermal',
        'output_name': 'temperature',
    },
    {
        'plot_group': 'Magnetic Flux Density (mf)',
        'output_name': 'magnetic_flux_density',
        'feature': 'Volume 1',
        'unit': 'mT',
        'color_range': (0, 100),
    },
]

outputs = msk.export_comsol_3d_plots(
    model_file=Path('E_quartercore_thermal_solved.mph'),
    plots=plots,
    view_plane=view_plane,
    dataset_name='Study (EM + Thermal)//Solution 1',
    image_size_px=(1400, 900),
)
for field, output_file in outputs.items():
    print(f'{field}: {output_file}')
```
