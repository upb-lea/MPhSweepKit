```python
import mph
import mphsweepkit as msk
import numpy as np
```

Start the Comsol Server and load the model

```python
# Start the COMSOL client
client = mph.start()
```

```python
# Load the model
model = client.load('toroid_cross_section_initial.mph')
```

Create a CascadedSweepModel

```python
# Create a CascadedSweepModel object
csm = msk.CascadedSweepModel(model, 'Study on Cross-Sections', show_param_names=True)
```

Change the material to 3F46

```python
csm.set_material_sweep(sweep_name='Material Sweep', 
                       material_names=['matsw.comp1.core1'], 
                       material_values=[[2]], 
                       sweep_type="filled")

csm.input_data
```

Set geometric sweep parameters to get circles with radii: 5, 10, 15 mm

```python
    # Create arrays for the parameters based on the radii
h_c = np.array([10, 10, 10])  # height
d_o = np.array([14, 18, 22])  # outer diameter
d_i = np.array([10, 10, 10])  # inner diameter

# Set the parametric sweep for the 'Geometry Sweep'
csm.set_parametric_sweep(sweep_name='Geometry Sweep',
                         param_names=['h_c', 'd_o', 'd_i'], 
                         param_units=['mm', 'mm', 'mm'],
                         param_values=[h_c, 
                                       d_o, 
                                       d_i],
                         sweep_type="sparse")

csm.input_data
```

Simulation

```python
csm.simulate()
```

Save the model with solution data:

```python
model.save('toroid_cross_section_solved.mph')
```

Release the Client

```python
client.remove(model)
client.clear()
client.disconnect()
```
